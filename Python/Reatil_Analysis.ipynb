```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 

# Display all columns
pd.set_option("display.max_columns", None)

# Display numbers with 2 decimal places
pd.options.display.float_format = '{:.2f}'.format
```

```python
# Read Dataset
df = pd.read_csv("sample_superstore.csv", encoding="cp1252")
df
```

```python
# Basic information 
df.shape
df.info()
```

```python
# statistical summary of the data
df.describe()
```

```python
# columns
df.columns
```

```python
# data cleaning

# Rename columns

df.columns=df.columns.str.replace("-","_")
df.columns=df.columns.str.replace(" ","_")
df.columns=df.columns.str.lower()
df.columns
```

```python
# checking missing values 
df.isnull().sum()
```

```python
# duplicate records
df.duplicated().sum()
# no duplicates are here ,if in-case duplicates may enter in future data use this drop_duplicates()
df=df.drop_duplicates()
```

```python
# converting date columns
df['order_date']=pd.to_datetime(df['order_date'])
df['ship_date']=pd.to_datetime(df['ship_date'])

# check once datatypes
df.dtypes
```

```python
# creating new features
# shipping days

df['shipping_days']=(df['ship_date']-df['order_date']).dt.days
```

```python
# extracting year,month,quarter,weekday from order_Date
df['year']=df['order_date'].dt.year
df['month']=df['order_date'].dt.month
df['quarter']=df['order_date'].dt.quarter
df['weekday']=df['order_date'].dt.weekday
```

```python
# Profit Margin 
df['profit_margin']=(df['profit']/df['sales'])*100
```

```python
# Loss Status 

df['loss_status']=np.where(df['profit']<0,'loss','profit')
```

```python
# Discount Category
df['discount_category']=pd.cut(df['discount'],
                               bins=[-1,0,0.2,0.5,1],
                               labels=['No Discount', 'Low','Medium','High'])
```

```python
# Revenue Category
df['revenue_category']=pd.cut(df['sales'],bins=[0,500,1000,25000],
                              labels=['low','medium','high'])
```

```python
# Exploratory Data Analysis ( EDA )

# Total Sales 

print("Total Sales :", round(df['sales'].sum(),2))
```

```python
# Total Profit 
print("Total Profit:",round(df['profit'].sum(),2))
```

```python
# Total Orders
print("Total Orders:",df['order_id'].nunique())
```

```python
# Total Customers
print("Total Customers:",df['customer_id'].nunique())
```

```python
# Total Products 
print("Total Products:",df['product_id'].nunique())
```

```python
# Average Order Value
print("Average Order Value:",round(df['sales'].mean(),2))
```

```python
# Average Discount

print("Average Discount :",round(df['discount'].mean(),2))
```

```python
# Average Shipping Days
print("Average Shipping Days:",round(df['shipping_days'].mean(),2))
```

```python
# Business Analysis

# sales by Category
df.groupby("category")['sales'].sum().sort_values(ascending=False)
```

```python
# Profit by Category
df.groupby('category')['profit'].sum().sort_values(ascending=False)
```

```python
# Sales by Region
df.groupby("region")['sales'].sum().sort_values(ascending=False)
```

```python
# profit by region
df.groupby('region')['profit'].sum().sort_values(ascending=False)
```

```python
# Top customers 
print("Top 10 Customers : ")
df.groupby('customer_name')['sales'].sum().sort_values(ascending=False).head(10)
```

```python
# Top 10 products 
print("Top 10 products : ")
df.groupby('product_name')['sales'].sum().sort_values(ascending=False).head(10)
```

```python
# Bottom 10 products 
print("Bottom 10 products : ")
df.groupby('product_name')['sales'].sum().sort_values(ascending=False).tail(10)
```

```python
# Monthly Sales 
monthly_sales=(df.groupby(df['order_date'].dt.to_period("M"))['sales'].sum())
monthly_sales
```

```python
# visualizations 

# Monthly Sales Trend
monthly_sales.plot(figsize=(12,6),marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
#Sales by category
df.groupby('category')['sales'].sum().plot(kind='bar',figsize=(8,5))
plt.title("Sales by Category")
plt.ylabel("Sales")
plt.show()
```

```python
# profit by region
df.groupby("region")['profit'].sum().plot(kind='bar',figsize=(8,5))
plt.title("Profit by Region")
plt.ylabel("profit")
plt.show()
```

```python
# Top Customers
df.groupby("customer_name")['sales'].sum().nlargest(10).plot(kind='barh')
plt.title(" Top 10 Customers ")
plt.xlabel("Sales")
plt.show()
```

Business Insights : 

-> Technology generates the highest sales.

-> Some Furniture products have negative profits despite high sales.

-> The West region contributes the highest revenue.

-> Heavy discounts reduce profitability in several product categories.

-> A small group of customers contributes a significant share of total sales.

-> Standard Class is the most commonly used shipping mode.
